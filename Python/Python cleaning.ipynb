```python
age_clean = df_age.copy()
months_clean = df_months.copy()
gender_clean = df_gender.copy()
road_user_clean = df_road_user.copy()
vehicle_clean = df_vehicle.copy()
ambulance_clean = df_ambulance.copy()
population_clean = df_population.copy()
```

```python
clean_data = {
    "Age Group": age_clean,
    "Months": months_clean,
    "Gender": gender_clean,
    "Road User Type": road_user_clean,
    "Vehicle Type": vehicle_clean,
    "Ambulance": ambulance_clean,
    "Population": population_clean
}
```

```python
for name, df in clean_data.items():
    df.columns = (
        df.columns
        .str.strip()
        .str.replace(" ", "_")
    )
```

```python
for name, df in clean_data.items():
    print(name)
    print(df.columns.tolist())
    print()
```

```python
for name, df in clean_data.items():
    text_columns = df.select_dtypes(include="object").columns

    for col in text_columns:
        df[col] = df[col].str.strip()
```

```python
for name, df in clean_data.items():
    print("=" * 50)
    print(name)

    for col in df.select_dtypes(include="object").columns:
        spaces = df[col].astype(str).str.contains(r"^\s|\s$", regex=True).sum()
        print(col, ":", spaces)
```

```python
for name, df in clean_data.items():
    print("=" * 50)
    print(name)

    numeric_columns = df.select_dtypes(include="number").columns

    for col in numeric_columns:
        negative_count = (df[col] < 0).sum()
        print(f"{col}: {negative_count} negative values")
```

```python
numeric_data = {
    "Age Group": df_age,
    "Months": df_months,
    "Gender": df_gender,
    "Road User Type": df_road_user,
    "Vehicle Type": df_vehicle,
    "Ambulance": df_ambulance,
    "Population": df_population
}

print("DataFrames loaded successfully!")
```

```python
for name, df in numeric_data.items():

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    for col in df.select_dtypes(include="number").columns:

        # Skip Year
        if col == "Year":
            continue

        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[
            (df[col] < lower_bound) |
            (df[col] > upper_bound)
        ]

        print(f"\n{col}")
        print(f"Q1 = {Q1}")
        print(f"Q3 = {Q3}")
        print(f"IQR = {IQR}")
        print(f"Lower Bound = {lower_bound}")
        print(f"Upper Bound = {upper_bound}")
        print(f"Outliers = {len(outliers)}")
```

```python
for name, df in numeric_data.items():

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    for col in df.select_dtypes(include="number").columns:

        if col == "Year":
            continue

        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[
            (df[col] < lower_bound) |
            (df[col] > upper_bound)
        ]

        if len(outliers) > 0:

            print(f"\n--- {col} ---")

            print(
                outliers[
                    ["Governorate", "Year", col]
                ].sort_values(
                    by=col,
                    ascending=False
                ).head(10)
            )
```

```python
for name, df in numeric_data.items():

    numeric_cols = [
        col for col in df.select_dtypes(include="number").columns
        if col != "Year"
    ]

    for col in numeric_cols:

        plt.figure(figsize=(8, 4))

        sns.boxplot(x=df[col])

        plt.title(f"{name} - {col} Outlier Detection")
        plt.xlabel(col)

        plt.show()
```
